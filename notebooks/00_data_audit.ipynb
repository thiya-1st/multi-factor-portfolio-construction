```python
sample_tickers = {
    "RELX": "REL.L",
    "Experian": "EXPN.L",
    "Sage": "SGE.L",
    "Unilever": "ULVR.L",
    "Diageo": "DGE.L",
    "Halma": "HLMA.L",
    "Intertek": "ITRK.L",
    "AstraZeneca": "AZN.L",
    "GSK": "GSK.L",
    "Bytes Technology": "BYIT.L",   
    "Games Workshop": "GAW.L",      
}
```

```python
results = []

for name, ticker_symbol in sample_tickers.items():
    ticker = yf.Ticker(ticker_symbol)
    info = ticker.info

    prices = ticker.history(period = "5y")
    years_of_price_history = len(prices)

    has_roe = info.get("returnOnEquity") is not None

    financials = ticker.financials
    has_financials = len(financials) > 0
    years_of_financials = financials.shape[1] if has_financials else 0

    results.append({
        "Company": name,
        "Ticker": ticker_symbol,
        "Years of Price History": years_of_price_history,
        "Has Current ROE": has_roe,
        "Has Historical Financials": has_financials,
        "Years of Financials": years_of_financials,
    })

yfinance_data_audit = pd.DataFrame(results)
yfinance_data_audit
```

## Data Audit Findings — yfinance

**Sample tested:** 11 FTSE 350 companies across sectors (info services, software, 
consumer staples, industrials, healthcare) and sizes (large-cap to small-cap).

**Results:**
- Price history: 11/11 companies had ~5 years of daily price data (~1,262 rows). 
  Consistent across large and small caps — no gaps found.
- Current fundamentals (ROE snapshot): 11/11 companies returned a value.
- Historical financials: 11/11 companies had historical financial statements 
  available, covering 4-5 years.

**Limitations found:**
1. **History depth is limited to ~5 years.** The project originally aimed to 
   backtest from ~2013, but yfinance's historical financials only go back 
   4-5 years. This may require either a paid data source for deeper history, 
   or narrowing the backtest period.
2. **No filing/report dates available.** yfinance only provides fiscal 
   period-end dates (e.g. "2023-12-31"), not the date results were actually 
   published. This creates a look-ahead bias risk — without filing dates, 
   we cannot be certain fundamentals were actually public knowledge at the 
   point they'd be used in the backtest. A fallback assumption (e.g. a fixed 
   ~90-120 day reporting lag) may be needed if a better data source doesn't 
   solve this.

**Conclusion:** yfinance is reliable and consistent for breadth (works across 
company size/sector), but insufficient on its own for depth and point-in-time 
accuracy. Next step: test Financial Modeling Prep (FMP) on the same sample to 
see if it resolves either limitation.
