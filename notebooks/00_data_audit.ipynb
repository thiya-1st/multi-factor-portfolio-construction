```python
sample_tickers = {
    "RELX": "REL.L",
    "Experian": "EXPN.L",
    "Sage": "SGE.L",
    "Unilever": "ULVR.L",
    "Diageo": "DGE.L",
    "Halma": "HLMA.L",
    "Intertek": "ITRK.L",
    "AstraZeneca": "AZN.L",
    "GSK": "GSK.L",
    "Bytes Technology": "BYIT.L",   
    "Games Workshop": "GAW.L",      
}
```

```python
results = []

for name, ticker_symbol in sample_tickers.items():
    ticker = yf.Ticker(ticker_symbol)
    info = ticker.info

    prices = ticker.history(period = "5y")
    years_of_price_history = len(prices)

    has_roe = info.get("returnOnEquity") is not None

    financials = ticker.financials
    has_financials = len(financials) > 0
    years_of_financials = financials.shape[1] if has_financials else 0

    results.append({
        "Company": name,
        "Ticker": ticker_symbol,
        "Years of Price History": years_of_price_history,
        "Has Current ROE": has_roe,
        "Has Historical Financials": has_financials,
        "Years of Financials": years_of_financials,
    })

yfinance_data_audit = pd.DataFrame(results)
yfinance_data_audit
```

## Data Audit Findings — yfinance

**Sample tested:** 11 FTSE 350 companies across sectors (info services, software, 
consumer staples, industrials, healthcare) and sizes (large-cap to small-cap).

**Results:**
- Price history: 11/11 companies had ~5 years of daily price data (~1,262 rows). 
  Consistent across large and small caps — no gaps found.
- Current fundamentals (ROE snapshot): 11/11 companies returned a value.
- Historical financials: 11/11 companies had historical financial statements 
  available, covering 4-5 years.

**Limitations found:**
1. **History depth is limited to ~5 years.** The project originally aimed to 
   backtest from ~2013, but yfinance's historical financials only go back 
   4-5 years. This may require either a paid data source for deeper history, 
   or narrowing the backtest period.
2. **No filing/report dates available.** yfinance only provides fiscal 
   period-end dates (e.g. "2023-12-31"), not the date results were actually 
   published. This creates a look-ahead bias risk — without filing dates, 
   we cannot be certain fundamentals were actually public knowledge at the 
   point they'd be used in the backtest. A fallback assumption (e.g. a fixed 
   ~90-120 day reporting lag) may be needed if a better data source doesn't 
   solve this.

**Conclusion:** yfinance is reliable and consistent for breadth (works across 
company size/sector), but insufficient on its own for depth and point-in-time 
accuracy.

see if it resolves either limitation.

## Data Audit Findings — FMP (Financial Modeling Prep)

**Test:** Same income statement endpoint tested on RELX (REL.L, UK) vs AAPL (US, control test).

**Results:**
- REL.L (UK): HTTP 402 error — "Premium Query Parameter... not available under your current subscription." UK-listed tickers are not accessible on the free tier.
- AAPL (US): HTTP 200 success — full data returned, including 5 years of   history (2021-2025) and a `filingDate` field separate from the fiscal period-end date (e.g. period end "2025-09-27" vs filing date "2025-10-31") — solving the point-in-time/look-ahead bias issue that yfinance could not.

**Conclusion:** 
FMP's data structure is genuinely better than yfinance's — it includes real filing dates, which is critical for avoiding look-ahead bias. However, the free tier only covers US-listed companies. Since this project's universe is FTSE 350 (UK), FMP's free tier is not usable as-is. A paid FMP plan with international coverage would be required to use this source, or an institutional source (Capital IQ / LSEG Datastream via Warwick) is a better fit given it's free through the university.

**Next step:** Test Capital IQ (via Warwick access) on the same sample to check whether it resolves both remaining gaps — UK coverage and filing dates.

## Data Source Decision — yfinance (Phase 1)

**Why not Capital IQ Pro:** Capital IQ Pro was tested and found to have excellent data — full FTSE 350 coverage, deep historical range (RELX data visible back to the 1990s), and explicit filing dates on as-reported filings, solving the two biggest gaps found in yfinance and FMP. However, Excel plugin 
login access could not be resolved (authentication issue between web login and desktop plugin), blocking programmatic/bulk data extraction needed for a 350-company pipeline. Manual pulls for hundreds of companies are not feasible. Capital IQ remains a planned extension if access is resolved later.

**Decision:** Proceed with yfinance for the initial pipeline build.

**Known limitations and how they'll be handled:**
1. **~5 years of historical fundamentals only** (vs. desired 2013+ range). 
   Accepted for Phase 1 — backtest window will be scoped to available history. Extending to a longer window is a planned future step, either via Capital IQ (if access is fixed) or a paid data source.
2. **No filing/report dates** — only fiscal period-end dates available. 
   Addressed with a fixed reporting-lag assumption: fundamentals are treated as "known" 90 days after fiscal period-end, approximating UK reporting deadlines (interim results due within 3 months, annual results within 4 months of period-end). This is a documented simplification, not a perfect fix, and will be flagged as a limitation in the final memo.

**Design note:** the data pipeline will be built so the *source* is a swappable component — if a better data source becomes available later, only the data-fetching layer should need to change, not the downstream factor scoring or backtesting logic.
